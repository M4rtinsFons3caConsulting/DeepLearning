# Album Augmenter

This notebook is used to run the album augmentation routines.

## Imports

Importing the script.

```python
import pandas as pd
from pathlib import Path

from deep.constants import METADATA_FILE, METADATA_DIR, UPSAMPLE_JSONS
from deep.preprocess import album_augmenter
```

```python
# Configuration values
label = "is_animal"
min_sample = 6000
target_ratio = 2
output_name = "binary_upsample"
```

## Generate or load an Upsample map

Images are mapped to transformations, until a desired number of samples is reached.

```python
# Read the relevant data, selecting only the columns of interest
# df = pd.read_csv(METADATA_FILE)[["rare_species_id", "file_path", f"{label}"]]


# For the binary case we need to concatenate
from deep.constants import METADATA_DIR
df = pd.concat([
    pd.read_csv(METADATA_FILE)[["rare_species_id", "file_path", label]],
    pd.read_csv(METADATA_DIR / "cropped_labels.csv")[["rare_species_id", "file_path", label]]
], ignore_index=True)
```

```python
# A plan can be generated
plan = album_augmenter.generate_oversample_map(
    df=df,
    label=label,
    target_ratio=target_ratio,
    min_samples=min_sample
)
plan = Path(UPSAMPLE_JSONS/plan)

# or, read from the location
# path = UPSAMPLE_JSONS / "oversample_plan_20250426T141041Z.json"
# plan = path
```

## Apply the map

Applying the map will generate the necessary images. 

```python
album_augmenter.oversample_labels(
    output_name=output_name,
    label=label,
    plan=plan
)
```
